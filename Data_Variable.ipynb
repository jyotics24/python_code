```python
# Float Data Type
# float data ttype is used to allocate memory for float value. (12.5)
f1=1.1
f2=12.4
f3=45.232324
print(type(f1),type(f2),type(f3))
f1,f2,f3
```

```python
# Complex Data type
# This data typt is used for reserve memory for complex number. ( real & imaginary )
# Complex number having two properties one is real value and another is imaginary value.
c1=10+22j
print(c1)
print(type(c1))
print(c1.real,": Is Real Value")
print(c1.imag,": Is imaginary Value")
```

```python
# Boolean Data type.
# This data type is used to reserve memory for Boolean Value. ( True & False )
# Boolean are represented using two keyword.
# True - 1
# False - 0
B1=True
B2=False
print(type(B1),type(B2))
print(B1,B2)
```

```python
# None Data type
# This data type is used to reserve memory for None value. ( None )
N=None
print(type(N))
print(N)
```

```python
# String
# str()
# This function is used to convert other type object to string.
print("------ Before Convert ------")
print()
int=12
print("Value Is :",int,", Type Is :",type(int))
float=12.50
print("Value Is :",float,", Type Is :",type(float))
complex=1+2j
print("Value Is :",complex,", Type Is :",type(complex))
bool1=True
print("Value Is :",bool1,", Type Is :",type(bool1))
bool2=False
print("Value Is :",bool2,", Type Is :",type(bool2))
print()
print("------ After Convert ------")
print()
int=str(12)
print("Value Is :",int,", Type Is :",type(int))
float=str(12.50)
print("Value Is :",float,", Type Is :",type(float))
complex=str(1+2j)
print("Value Is :",complex,", Type Is :",type(complex))
bool1=str(True)
print("Value Is :",bool1,", Type Is :",type(bool1))
bool2=str(False)
print("Value Is :",bool2,", Type Is :",type(bool2))
```
