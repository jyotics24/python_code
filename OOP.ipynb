```python
# Create a car class with attibute like brand & model, Then create a instance of class
class car:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
my_car=car("TATA","Safari") # Create an instance of the Car class
print(my_car.brand)
print(my_car.model)
                                                                
```

```python
# Add method to the car class that display the full name of car.
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def full_name(self):
        return f"{self.model} {self.brand}"

# Create an instance of the Car class
my_car = Car("TATA", "Safari")

print(my_car.brand)
print(my_car.model)
print(my_car.full_name()) # use () symbol in print
```

```python
# Inheritance
# Create a Elictric car class that inherits from the car class and has an additional attribute battry size.
class car:
    def __init__(self,brand,model):
        self.brand=f'Brand Is: {brand}'
        self.model=f'Model Is: {model}'
    def full_name(self):
        return f'Model Is: {self.model} | Brand Is: {self.brand}'
class electric_car(car):  # Define the subclass 'ElectricCar' that inherits from 'Car'
    def __init__(self, brand, model,battry_size):
								super().__init__(brand, model) # Call the parent class's constructor to initialize brand and model
								self.battry_size = f'Battery Size Is : {battry_size}' 
								
my_car= electric_car("TATA","CURVE_Ev","86 kWh")
print(my_car.brand)
print(my_car.model)
print(my_car.battry_size)
print(my_car.full_name())
```

```python
# Encapsulation
# Modify the car class to encapsulate the brand attribute, making it private and provide a getter method for it.
# A getter method is used to access the value of a private attribute from outside the class.
class car:
    def __init__(self,brand,model):
       self.__brand=f'Brand Is: {brand}'
       self.model=f'Model Is: {model}'
    def get_brand(self):
        return self.__brand + " ! " # Just a symbol 
    def full_name(self):
        return f'{self.__brand} | {self.model}'
class electri_car(car):
    def __init__(self, brand, model,battery):
        super().__init__(brand, model)
        self.battery=f'Battery Power Is: {battery}'
my_car=electri_car("TATA","Curve_Ev","85 kWh")
# print(my_car.__brand) # Error
print(my_car.get_brand()) # we can only use function name
# print(my_car.brand) # not use
print(my_car.model)
print(my_car.battery)
print(my_car.full_name())
```

```python
# Polymorphism
# Demonstrate polymorphism by defining a method fuel_type() in both Car and ElectricCar classes, where each class provides a different implementation.
class Car:
    def __init__(self, brand, model):
        self.__brand = f'Brand Is: {brand}'
        self.model = f'Model Is: {model}'
    
    def get_brand(self):
        return self.__brand + " !"  # just a symbol
        
    def full_name(self):
        return f'{self.__brand} | {self.model}'
    
    def fuel_type(self):
        return f'{self.__brand} | {self.model} : Petrol Or Diesel'

class Electric_Car(Car):
    def __init__(self, brand, model, battery):
        super().__init__(brand, model)
        self.battery = f'Battery Power Is: {battery}'

    def fuel_type(self):
        return f"{self.get_brand()} | {self.model} : Electric Charge" #  Use Self.get_brand() Because it a encapsulation
    
# Creating an object of Electric_Car
my_Car = Electric_Car("TATA", "Curve_Ev", "85 kWh")
# Print fuel type of electric Car
print(my_Car.fuel_type()) 
# Creating an object of regular petrol/diesel Car
p_Car = Car("TATA", "Safari")
# Print fuel type of traditional Car
print(p_Car.fuel_type())
```

```python
# class variable 
# Add a class variable to car that keeps the track the number of car create
class Car:
    total_car = 0  # Class variable to track number of Car instances
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        Car.total_car += 1  # Increment the class variable
Car("Tata", "Safari")
Car("Suzuki", "Swift")
print(Car.total_car) # Total number of cars created
```

```python
# static method
# Add a static method to the Car class that returns a general description of a car
class car:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    @staticmethod
    def general_info():
        return " Car are used on Transport"
print(car.general_info()) # we don't need to create a object bcz is a we use static method
```
