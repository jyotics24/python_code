```python
# Slicing

# Index is used to read a single value from a sequence (like list, string, or tuple).
# Slicing is used to read more than one value (a range of values) from a sequence.

# Slicing is done in two ways:
# 1. Slice Operator
# 2. Slice data type or slice object

# ----------------------------------------------------
# 1. Slice Operator
# ----------------------------------------------------
# Slice Operator takes start, stop, and step values.
# If step is positive:
#     - Default start = 0
#     - Default stop = length of sequence
# If step is negative:
#     - Default start = -1 (last index)
#     - Default stop = -(length + 1) (before first index)

# Syntax: sequence_name[start : stop : step]

# Example 1: Full slice using default start, stop, step
my_list = [10, 20, 30, 40, 50]
result = my_list[::]  # returns all elements
print("Ex-1:", result) 

# Example 2: Slicing with positive step
result = my_list[1:4:1]  # from index 1 to 3
print("Ex-2:", result) 

# Example 3: Slicing with step of 2
result = my_list[::2]
print("Ex-3:", result)

# Example 4: Reverse the list using negative step
result = my_list[::-1]
print("Ex-4:", result) 

# Example 5: Slice with negative step and range
result = my_list[4:1:-1]
print("Ex-5:", result)
```

```python
# 2. Slice Data Type or Slice Object
# Python has a built-in function called slice()
# It is used to create a slice object that we can use with lists, strings, etc.
# Syntax: slice(start, stop, step)

# Example 1: Use slice object to get some elements
nums = [10, 20, 30, 40, 50, 60]
my_slice = slice(1, 5, 2)     # same as nums[1:5:2]
print("Ex-1:", nums[my_slice])

# Example 2: slice with start and stop only
part = slice(2, 5)            # same as nums[2:5]
print("Ex-2:", nums[part])

# Example 3: slice with only stop
first_part = slice(3)         # same as nums[0:3]
print("Ex-3:", nums[first_part])

# Example 4: slice every second item
skip_one = slice(0, len(nums), 2)
print("Ex-4:", nums[skip_one])
```

```python
# Write a program to sort element of array.(list)
# Using Bubble Sort
list=[]
n=int(input("Enter The List Size :"))
for i in range(n):
    value=int(input("Enter No In List :"))
    list.append(value)
print(f'Before Sorted List: {list}')
for j in range(n):
    for i in range(n-1):
        if list[i]>list[i+1]:
            list[i+1],list[i]=list[i],list[i+1]
print(f'After Sorted List: {list}')
```

```python
numbers = []
n = int(input("Enter how many numbers: "))

for i in range(n):
    num = int(input(f"Enter number {i + 1}: "))
    numbers.append(num)

print("\nBefore sorting:", numbers)
print("\n--- Starting Bubble Sort Dry Run ---")
# Bubble Sort with dry run
for i in range(n):
    print(f"\nPass {i + 1}:")
    for j in range(n - 1):
        print(f"  Comparing {numbers[j]} and {numbers[j + 1]}", end=" ")
        if numbers[j] > numbers[j + 1]:
            # Swap if the left number is bigger
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
            print(f"=> swapped ->{numbers}")
        else:
            print("=> no change")
    print(f"  Result after pass {i + 1}: {numbers}")

print("\n--- Bubble Sort Completed ---")
print("After sorting:", numbers)
```

```python
# insert method()
# The insert() method is used to insert a single element at a specified index in the list.
# Syntax: list_name.insert(index, value)
# - 'index' is the position in the list where the value should be inserted.
# - 'value' is the item you want to insert.

# Ex-1
list_1 = [10, 20, 40, 50]
list_1.insert(2, 30)  # Insert 30 at index 2
print("After insert():", list_1) 
```

```python
# This method never gives an error if the given index is invalid.

# Example 1: Index greater than list length
list_1 = [1, 2, 3]
list_1.insert(100, 4)  # 100 is out of range
print(list_1)  # [1, 2, 3, 4] - inserted at the end

# Example 2: Very negative index
list_2 = [10, 20, 30]
list_2.insert(-100, 5)  # -100 is out of range on the negative side
print(list_2)  # Output: [5, 10, 20, 30] - inserted at the start
```

```python
# inserting multiple value or more than one value using slicing
# Use slicing to insert or replace multiple values
# syntax : list-name [start:stop:step]=[value1,value2,value3,...]

# Insert multiple values at a specific index
list_2 = [10, 20, 50, 60]
list_2[2:2] = [30, 40]  # Insert 30 and 40 at index 2
print("Insert multiple using slicing:", list_2)  # Output: [10, 20, 30, 40, 50, 60]

# Replace multiple elements
list_3 = [1, 2, 3, 4, 5]
list_3[1:3] = [10, 20]  # Replace 2 and 3 with 10 and 20
print("Replace with slicing:", list_3)
```

```python
# Program to perform append, insert, and update operations on a list.

ope = []

n = int(input("Enter number of operations: "))

for i in range(n):
    cmd = input("Enter command (append, insert, update): ").strip().lower()
    
    if cmd == "append":
        value = int(input("Enter value: "))
        ope.append(value)
        
    elif cmd == "insert":
        index = int(input("Enter position: "))
        value = int(input("Enter value: "))
        ope.insert(index, value)
        
    elif cmd == "update":
        index = int(input("Enter position: "))
        value = int(input("Enter value: "))
        if 0 <= index < len(ope):
            ope[index] = value
        else:
            print("Invalid index")
    
    else:
        print("Invalid command")

print("Final list:", ope)
```

```python
# Using del keyword
# 'del' keyword is used to delete one or more than one element
# 'del' keyword required 2 things
# 1. index // we can delete one value, if given index is not valid, it raises IndexError
# 2. slicing // we can delete more than one value.
```

```python

# Ex-1 Deleting one element by index
list_1 = [10, 20, 30, 40, 50]
del list_1[2]  # Deletes the element at index 2 (30)
print("After deleting index 2:", list_1)  # Output: [10, 20, 40, 50]

# Ex-2 Deleting multiple elements using slicing
list_2 = [100, 200, 300, 400, 500, 600]
del list_2[1:4]  # Deletes elements from index 1 to 3 (200, 300, 400)
print("After deleting from index 1 to 3:", list_2)  # Output: [100, 500, 600]
```

```python
# Using pop method
# We can use a list as a stack data structure.
# Stack follows LIFO (Last In, First Out) principle.
# We can implement a stack using a list.
# 1. Push - Add element (append).
# 2. Pop - Remove and return the last element (pop).

# Example 1: Push (append)
stack = []
print("Initial stack:", stack)

stack.append(10)
stack.append(20)
stack.append(30)
print("Stack after pushes:", stack)

# Example 2: Pop (remove last element)
r = stack.pop()
print("Popped element:", r)
print("Stack after pop:", stack)
```

```python
# Write a program to perfrom push,pop and display operation on stack.
n=int(input("Enter No of Operation"))
stack=[]
for i in range(n):
    cmd=input("Choose the command (push,pop,display)")
    if cmd=="push":
        value=int(input("Enter The Value"))
        stack.append(value)
        print(f'{value} Pushed Inside Stack')
    elif cmd=="pop":
        if len(stack)==0:
            print("Stack Is Empty")
        else:
            value=stack.pop()
            print(f'{value} Poped from stack')
    elif cmd=="display":
        print("The Stack Is :",stack)
```
