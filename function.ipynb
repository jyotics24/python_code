```python
def test_case(): # calling function
    print("Start Function")
if(__name__=="__main__"):
        test_case() # function call
```

```python
# formal & actual 
def product(name,price,exp_date): # formal parameter
    print("Product Name :",name)
    print("product Price :", price)
    print("product Exp-Date :",exp_date)
if(__name__=="__main__"):
    product("Ice",100,2025) # actual parameter
```

```python
product("Rice",1200,2026)
```
