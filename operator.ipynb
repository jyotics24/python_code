# OPERATOR

```python
# What is operator 
# Operator is a symbol that performs an operation on variables and values.
# Operator are classified in 3 type's
# 1. Unary Operator ( Perfrom operation on one operand )
# 2. Binary Operator ( Perfron operation on two operand )
# 3. Ternary Operator ( Perfrom operation on three operand )
```

```python
# Types Of operator 
# 1. Arithmetic Operator
# 2. Relational Operator
# 3. Conditional Operator
# 4. Bitwise Operator
# 5. Assignment Operator
# 6. Logical Operator
# 7. Membership Operator
# 8. Identity Operator
```

```python
# Operator ( % )
# This Operator divide two number and return remainder.
print(6%2)
print(5%2)
```

```python
# Operator ( ** )
# This Operator used to raise a number (the base) to the power of another number (the exponent).
a=4**2
print(a)
p=6**3
print(p)
```

```python
# Write a program to check number is even or odd.
num=int(input("Enter The Number :"))
if num%2==0:
    print(f"{num} -> Is Even")
else:
    print(f"{num} -> Is Odd")
```

```python
# Write a program to find last digit of input number is even or odd
# 123 --> Check last No 3 is even or odd
# 422 --> Check last No 2 is even or odd
# How i get last Number
# 123%10=3
# 422%10=2
num=int(input("Enter The No :"))
print("The Entered No Is :",num)
l=num%10
print("The Last No Is :",l)
print()
if l%2==0:
    print(l,"- Is Even")
else:
    print(l,"- Is Odd")
```

```python
# Write a program to find max of two number.
n1=int(input("Enter the 1st Number :"))
n2=int(input("Enter The 2nd Number :"))
print("First Number Is :",n1)
print("Second Number Is :",n2)
if n1>n2:
    print(n1," - First Number Is Max")
else:
    print(n2," - Second Number Is Max")
```

```python
# Write a program to input the name and age of a person to check they eligible to vote or not.
name=input("Enter the name :")
age=int(input("Enter the age to check Eligible for Vote :"))
print("Name Is :",name)
print("Age Is ",age)
print()
if age>=18:
    print(f"{name} Eligible For Vote.")
else:
    print(f"{name} Not Eligible For Vote.")
```
